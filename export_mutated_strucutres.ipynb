```python
import numpy as np

import matplotlib.pyplot as plt

import pandas as pd
import os
import sys

from pathlib import Path


from oxDNA_analysis_tools.UTILS.oxview import oxdna_conf, from_path
from oxDNA_analysis_tools.UTILS.RyeReader import describe, get_confs

from ipy_oxdna.oxdna_simulation import Simulation

%load_ext autoreload
%autoreload 2
```

```python
file_dir = Path('./science_melting/mutated_strucutres')

sim = Simulation(file_dir)

with open('mutate_structring_strand.js', 'r') as f:
    mutate_structring_strand = f.readlines()

my_path = f'{file_dir.absolute()}/trial_0'

mutation_fractions = np.linspace(0.2, 0.8, 4)
mutation_algo = ['interspaced', 'start']
nucs_per_subregion = [5, 10, 15]

for algo in mutation_algo:
    for nucs in nucs_per_subregion:
        for fraction in mutation_fractions:
            
            sys_path = f'{my_path}/{algo}_{nucs}_{fraction}'
            os.makedirs(sys_path, exist_ok=True)
            
            mutate_structring_strand = [line.replace('ADD_YOUR_PATH_HERE', sys_path) if 'ADD_YOUR_PATH_HERE' in line else line for line in mutate_structring_strand]
            mutate_structring_strand = [line.replace('MUTATION_FRACTION', str(fraction)) if 'MUTATION_FRACTION' in line else line for line in mutate_structring_strand]
            mutate_structring_strand = [line.replace('ALGO', str(algo)) if 'ALGO' in line else line for line in mutate_structring_strand]
            mutate_structring_strand = [line.replace('SUBREGION', str(nucs)) if 'SUBREGION' in line else line for line in mutate_structring_strand]

            with open(f'{sys_path}/mutate_structring_strand.js', 'w') as f:
                f.writelines(mutate_structring_strand)
            break
        break
    break


top_info, traj_info = describe(sim.sim_files.top, sim.sim_files.dat)
conf = get_confs(top_info, traj_info, 0, 1)[0]


oxdna_conf(top_info, conf, script_file_path=f'{sys_path}/mutate_structring_strand.js')
```
